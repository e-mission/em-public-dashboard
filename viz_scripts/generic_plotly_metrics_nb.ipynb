```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd
    
# Row bar plots assume percentages
def make_single_row_panel(modes, percents, colors, panel_label, legend_group, legend_group_title):
    traces = []
    for mode, pct in zip(modes, percents):
        traces.append(
            go.Bar(
                y=[panel_label],
                x=[pct],
                orientation='h',
                marker=dict(
                    color=colors.get(mode, "#999"),
                    line=dict(color='black', width=1)
                ),
                name=str(mode),
                hovertemplate=f"{mode}: {pct:.1%}<extra></extra>",
                legendgroup=legend_group,
                legendgrouptitle_text=legend_group_title,
                text=[f"{pct:.1%}"],
                textposition="inside",
                insidetextanchor="middle",
                insidetextfont=dict(family='Times', size=20, color='black')
            )
        )
    return traces
```

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd

# Column bar plots print raw values, rather than percentages
def make_single_column_panel(modes, percents, colors, panel_label, legend_group, legend_group_title):
    traces = []
    for mode, pct in zip(modes, percents):
        traces.append(
            go.Bar(
                x=[panel_label],         # one column per panel
                y=[pct],                 # height = percent
                orientation='v',
                marker=dict(
                    color=colors.get(mode, "#999"),
                    line=dict(color='black', width=1)
                ),
                name=str(mode),
                hovertemplate=f"{mode}: {round(pct)}<extra></extra>",
                legendgroup=legend_group,
                legendgrouptitle_text=legend_group_title,
                text=[round(pct)],
                textposition="inside",
                insidetextanchor="end",
                insidetextfont=dict(family='Times', size=20, color='black')
            )
        )
    return traces
```

```python
def prepare_counts(df, mode_col, distance_col, threshold=0.02):
    counts = (
        df.groupby(mode_col)
          .agg({distance_col: "count"})
          .rename(columns={distance_col: "count"})
          .sort_values("count", ascending=False)
    )

    total = counts["count"].sum()

    if total == 0:
        return ["Other"], [1.0], [0]

    large_categories = {}
    other_count = 0

    for mode, count in counts["count"].items():
        proportion = count / total

        if proportion < threshold:
            other_count += count
        else:
            large_categories[mode] = count

    if other_count > 0:
        large_categories["Other"] = other_count

    modes = list(large_categories.keys())
    trip_counts = list(large_categories.values())
    percents = [c / total for c in trip_counts]

    return modes, percents, trip_counts


# Used to count distance in miles
def prepare_sums(df, group_col, distance_col, threshold=0.02):

    df_u80 = df

    # Group & count
    counts_by_group = (
        df_u80.groupby(group_col)[distance_col]
              .sum()
              .sort_values(ascending=False)
    )

    total = counts_by_group.sum()

    if total == 0:
        return ["Other"], [1.0], [0]

    large_categories = {}
    other_count = 0

    for mode, count in counts_by_group.items():
        proportion = count / total

        if proportion < threshold:
            other_count += count
        else:
            large_categories[mode] = count

    if other_count > 0:
        large_categories["Other"] = other_count

    modes = list(large_categories.keys())
    counts = list(large_categories.values())
    percents = [c / total for c in counts]

    return modes, percents, counts
```

```python
def filter_u80(df, distance_col, reference_df):
    """Filter to trips at or below the 80th percentile distance."""
    cutoff = reference_df[distance_col].quantile(0.8)

    if pd.isna(cutoff):
        cutoff = 0

    return df[df[distance_col] <= cutoff]
```

```python
def build_trip_plot(
    distance_col,
    panels,
    plot_title="",
    file_name="",
    threshold=0.00,
    apply_u80=False,
    layout=None,
    num_legends=1,
    use_distance=False,
    orientation="h",
    use_raw_values=False
):
    """
    Build a stacked multi-panel trip mode plot.
    `panels` must be a list of dictionaries. Each dict represents one panel.

    Required keys per panel:
        df          -> dataframe
        text        -> quality text description
        label       -> label for this panel ("User Labeled", "Inferred", etc.)
        legend_title-> legend group title
        mode_col    -> column for mode classification
        colors      -> color mapping for that panel
    """

    if not isinstance(panels, list) or len(panels) == 0:
        raise ValueError("`panels` must be a non-empty list of panel specifications.")

    required_keys = {"df", "text", "label", "legend_title", "mode_col", "colors"}
    for i, p in enumerate(panels):
        missing = required_keys - p.keys()
        if missing:
            raise ValueError(f"Panel {i+1} is missing required keys: {missing}")

    # Build subplot titles
    # subplot_titles = [
    #     f"{p['label']}<br>{p['text']}" for p in panels
    # ]

    subplot_titles = [
        f"{p.get('title', p['label'])}<br>{p['text']}" for p in panels
    ]

    fig = make_subplots(
        rows=len(panels),
        cols=1,
        shared_xaxes=True,
        vertical_spacing=0.1,
        subplot_titles=subplot_titles,
    )

    # Add each panel's traces
    for i, p in enumerate(panels, start=1):

        # Data processing steps
        if use_raw_values:
            # modes = p["df"][p["mode_col"]]
            # percents = p["df"][distance_col]
            modes = p["df"][p["mode_col"]].tolist()
            percents = p["df"][distance_col].tolist()
            counts = p["df"]["Count"].tolist() if "Count" in p["df"].columns else [None]*len(modes)
            
        elif use_distance:
            modes, percents, counts = prepare_counts(
                p["df"], p["mode_col"], distance_col, threshold
            )
        else:
            modes, percents, counts = prepare_sums(
                p["df"], p["mode_col"], distance_col, threshold
            )

        # Vertical or horizontal bar plots
        if orientation == "h":
            traces = make_single_row_panel(
                modes,
                percents,
                p["colors"],
                p["label"],
                legend_group=f"group_{p['label'].lower().replace(' ', '_')}",
                legend_group_title=p["legend_title"],
            )
        else:
            traces = make_single_column_panel(
                modes,
                percents,
                p["colors"],
                p["label"],
                legend_group=f"group_{p['label'].lower().replace(' ', '_')}",
                legend_group_title=p["legend_title"],
            )


        showlegend = p.get("show_legend", True)

        for t in traces:
            t.showlegend = showlegend
            fig.add_trace(t, row=i, col=1)


    # default laout applied if none provided
    if layout is None:
        layout = dict(
            height=350 * len(panels),
            barmode="stack",
            showlegend=True,
            legend=dict(
                orientation="v",
                yanchor="top",
                xanchor="left",
                valign="top",
                traceorder="grouped",
                itemsizing="constant",
            ),
            title=dict(text="PLOT TITLE", x=0.0),
        )
    else:
        fig.update_layout(layout)


    fig.update_xaxes(tickformat=".0%")
    output_name = f"{file_name}.html"
    fig.write_html(output_name, include_plotlyjs="cdn", full_html=True)

    return fig, output_name
```
